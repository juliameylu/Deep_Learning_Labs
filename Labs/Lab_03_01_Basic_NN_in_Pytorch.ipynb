1. Extend the code from Notebook 3.1 to implement an MLP with a single hidden layer of size 100.

```python
import numpy as np
import torch
from palmerpenguins import load_penguins
from matplotlib import pyplot as plt
from torch import nn

df = load_penguins()

# drop rows with missing values
df.dropna(inplace=True)

# get two features
X = df[['flipper_length_mm','bill_length_mm']].values

# convert species labels to integers
y = df['species'].map({'Adelie':0,'Chinstrap':1,'Gentoo':2}).values
X -= np.mean(X,axis=0)
X = torch.tensor(X).float()
y = torch.tensor(y).long()
```

```python
mlp_model = torch.nn.Sequential(
    torch.nn.Linear(2,100), # two inputs, three outputs
    torch.nn.ReLU(),
    torch.nn.Linear(100,3)
)
loss_fn = torch.nn.CrossEntropyLoss()
lr = 1e-2
opt = torch.optim.SGD(mlp_model.parameters(), lr=lr)
epochs = 100
for epoch in range(epochs):
    opt.zero_grad() # zero out the gradients

    z = mlp_model(X) # compute z values
    loss = loss_fn(z,y) # compute loss

    loss.backward() # compute gradients

    opt.step() # apply gradients

    print(f'epoch {epoch}: loss is {loss.item()}')
```

2. Write code to compute the accuracy of each model.


```python
linear_model = torch.nn.Sequential(
    torch.nn.Linear(2,3), # two inputs, three outputs
)
loss_fn = torch.nn.CrossEntropyLoss()
lr = 1e-2
opt = torch.optim.SGD(linear_model.parameters(), lr=lr)
epochs = 100
for epoch in range(epochs):
    opt.zero_grad() # zero out the gradients

    z = linear_model(X) # compute z values
    loss = loss_fn(z,y) # compute loss

    loss.backward() # compute gradients

    opt.step() # apply gradients

    print(f'epoch {epoch}: loss is {loss.item()}')
```

```python
mlp_z = mlp_model(X)
mlp_preds = torch.argmax(mlp_z, dim=1)
mlp_accuracy = torch.mean((mlp_preds == y).float())
mlp_accuracy
```

```python
linear_z = linear_model(X)
linear_preds = torch.argmax(linear_z, dim=1)
linear_accuracy = torch.mean((linear_preds == y).float())
linear_accuracy
```

3. Compare the MLP and linear models in terms of accuracy on the training set.

The MLP and linear model had similar training accuracies. Meaning that the flipper and bill length features may already separate the penguin species pretty well with a linear boundary. So the extra hidden layer did not improve much.
